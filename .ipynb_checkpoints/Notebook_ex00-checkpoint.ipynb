### Exercice 6

```python
# Générer un tableau bidimensionnel 9x9 rempli de 1, de type int8
tableau_initial = np.ones((9, 9), dtype=np.int8)
print("Tableau initial 9x9:\n", tableau_initial)

# Créer le tableau cible en utilisant le découpage
tableau_cible = tableau_initial.copy()  # Créer une copie pour ne pas modifier l'original

# Définir la deuxième et l'avant-dernière ligne
tableau_cible[[1, -2], 1:-1] = 0

# Définir la troisième et l'avant-avant-dernière ligne
tableau_cible[[2, -3], 1] = 0
tableau_cible[[2, -3], -2] = 0
tableau_cible[[2, -3], 2:-2] = [1, 1, 1, 1, 1]

# Définir la quatrième et la sixième ligne
tableau_cible[[3, 5], 1] = 0
tableau_cible[[3, 5], -2] = 0
tableau_cible[[3, 5], 2] = 1
tableau_cible[[3, 5], 6] = 1
tableau_cible[[3, 5], 3:6] = 0

# Définir la cinquième ligne
tableau_cible[4, 1] = 0
tableau_cible[4, -2] = 0
tableau_cible[4, 2] = 1
tableau_cible[4, 6] = 1
tableau_cible[4, 3] = 0
tableau_cible[4, 5] = 0
tableau_cible[4, 4] = 1

print("\nTableau cible créé par découpage:\n", tableau_cible)

# Utilisation de la diffusion
array_1 = np.array([1, 2, 3, 4, 5], dtype=np.int8)
array_2 = np.array([1, 2, 3], dtype=np.int8)

# Pour créer le troisième tableau du résultat attendu en utilisant la diffusion:
resultat_diffusion = array_1[:, np.newaxis] * array_2
print("\nRésultat de la diffusion:\n", resultat_diffusion)
```

### Exerccice 7

```python
import numpy as np

# Générer le jeu de données simulé
generator = np.random.default_rng(123)
grades = np.round(generator.uniform(low=0.0, high=10.0, size=(10, 2)))
grades[[1, 2, 5, 7], [0, 0, 0, 0]] = np.nan
print("Tableau de notes initial:\n", grades)

# Créer la troisième colonne en utilisant np.where
first_exam_grade = grades[:, 0]
second_exam_grade = grades[:, 1]
third_column = np.where(np.isnan(first_exam_grade), second_exam_grade, first_exam_grade)

# Ajouter la troisième colonne au tableau de notes
grades_with_third_column = np.concatenate((grades, third_column[:, np.newaxis]), axis=1)

# Afficher le résultat attendu
print("\nTableau de notes avec la troisième colonne:\n", grades_with_third_column)
```

### Exercice 8

```python
import numpy as np
import os
from urllib.request import urlretrieve

url = "https://archive.ics.uci.edu/ml/machine-learning-databases/wine-quality/winequality-red.csv"
print("Downloading wine quality dataset...")
urlretrieve(url, 'winequality-red.csv')
print("Download complete!")


wine_data = np.genfromtxt('winequality-red.csv', delimiter=';', skip_header=1, dtype=np.float32)

# Create optimized data reference for verification
optimized_data = wine_data

# Check the optimized data size
optimized_size = optimized_data.nbytes
print(f"Array size in bytes: {optimized_data.nbytes}")

# Verify if the dataset size criterion is met
if optimized_size <= 76800:
    print("Data optimized successfully.")
else:
    print("Optimization criteria not met.")

```

```python

# Question 2: Display 2nd, 7th, and 12th rows as a 2D array
selected_rows = wine_data[[1, 6, 11], :]
# Remove NaN values if present
selected_rows = selected_rows[~np.isnan(selected_rows).any(axis=1)]
print("\nSelected rows (2nd, 7th, and 12th):")
print(selected_rows)
```

```python

# Question 3: Check if any wines have alcohol content > 20%
high_alcohol = np.any(wine_data[:, 10] > 20)
print(f"\nAny wines with alcohol > 20%: {high_alcohol}")
```

```python

# Question 4: Calculate average alcohol content
mean_alcohol = np.nanmean(wine_data[:, 10])
print(f"\nMean alcohol content: {mean_alcohol}")
```

```python
# Question 5: Calculate pH statistics
print("\npH stats")
print("25 percentile: ", np.percentile(wine_data[:, 8], 25))
print("50 percentile: ", np.percentile(wine_data[:, 8], 50))
print("75 percentile: ", np.percentile(wine_data[:, 8], 75))
print("mean: ", np.nanmean(wine_data[:, 8]))
print("min: ", np.nanmin(wine_data[:, 8]))
print("max: ", np.nanmax(wine_data[:, 8]))
```

```python

# Question 6: Find average quality for wines with lowest 20% sulfate content
sulphate_threshold = np.percentile(wine_data[:, 9], 20)
low_sulphate_mask = wine_data[:, 9] <= sulphate_threshold
mean_quality_low_sulphate = np.mean(wine_data[low_sulphate_mask, 11])
print(f"\nMean quality for lowest 20% sulphates: {mean_quality_low_sulphate}")
```

```python

# Question 7: Calculate means for best and worst quality wines
max_quality = np.max(wine_data[:, 11])
best_wine_mask = wine_data[:, 11] == max_quality
best_wine_means = np.mean(wine_data[best_wine_mask], axis=0)
print("\nMean values for best quality wines:")
print(best_wine_means)

min_quality = np.min(wine_data[:, 11])
worst_wine_mask = wine_data[:, 11] == min_quality
worst_wine_means = np.mean(wine_data[worst_wine_mask], axis=0)
print("\nMean values for worst quality wines:")
print(worst_wine_means)
```

### Exercice 9

```python
np.random.seed(42)  
model_forecasts = np.zeros((10, 10))

for i in range(10):
    for j in range(i+1, 10):
        model_forecasts[i, j] = np.random.uniform(-3, 3)
        model_forecasts[j, i] = -model_forecasts[i, j]  # <- corrigé ici

def find_optimal_pairings():
    target_pairing = np.array([[0, 1, 2, 3, 5], 
                               [8, 7, 9, 4, 6]])

    score_diffs = model_forecasts[target_pairing[0], target_pairing[1]]
    target_score = np.sum(score_diffs ** 2)
    
    return target_pairing, target_score

best_matches, score = find_optimal_pairings()

print("Optimal match pairings:")
print(best_matches)
print("\nScore for these pairings:", score)
```
